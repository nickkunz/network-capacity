## Dependencies

```
## libraries
import sys
import matplotlib.pyplot as plt
from IPython.display import display
from pathlib import Path

## path
root = Path.cwd().resolve().parent
sys.path.insert(0, str(root))

## modules
from src.data.builders import load_processed_data
from src.estimators.factories import load_estimators
from src.visualizers import visualize

## constants
from src.evaluators.config import (
    FEAT_X, 
    FEAT_Z, 
    TARGET
)
```

```
## reproducibility
N_DECIMALS = 2
N_REPEATS = 30
RANDOM_STATE = 42

## paths
CACHE_DIR = root / "notebooks" / "cache"
FIGURE_DIR = root / "manuscript" / "latex" / "figures"

## analysis identity
data = load_processed_data()
models = load_estimators(random_state = RANDOM_STATE)
context = visualize.VisualizationContext(
    data = data,
    models = models,
    feat_x = tuple(FEAT_X),
    feat_z = tuple(FEAT_Z),
    target = TARGET,
    n_repeats = N_REPEATS,
    random_state = RANDOM_STATE,
    cache_dir = CACHE_DIR,
    figure_dir = FIGURE_DIR,
    n_decimals = N_DECIMALS,
)

print(f"Original Data: {data.shape[1]} features, {data.shape[0]} observations")
print(f"Learned Models: {len(models)} estimators")
```
